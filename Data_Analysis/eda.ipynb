#### **Merge and Concat** ####

```python
data1 = {
    'Name' : ['A', 'B', 'C'],
    'age' : [23, 45, 67],
    'salary' : [10000, 20000, 30000]
}

data2 = {
    'Name' : ['A', 'Bb', 'C'],
    'age' : [23, 54, 67],
    'salary' : [10000, 20002, 30000]
}
```

```python
new_df1 = pd.DataFrame(data1)
new_df2 = pd.DataFrame(data2)
```

```python
new_df1.head()
```

```python
new_df = pd.concat([new_df1, new_df2], ignore_index=True)
new_df.head()
```

```python
# pd.concat([new_df1, new_df2], ignore_index=True, axis=1)
```

```python
new_df1.drop_duplicates(inplace=True)
new_df.head()
```

```python
# SQL Join conditions using .merge() function

pd.merge(new_df1, new_df2, on='age', how='inner')
```
