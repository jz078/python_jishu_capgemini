```python
# 1️⃣ Basic Data Understanding
# load the dataset

import pandas as pd

df = pd.read_csv("tips.csv")
```

```python
# view first rows

df.head()
```

```python
# view last rows

df.tail()
```

```python
# shape of dataset

df.shape
```

```python
# column names

df.columns
```

```python
# data types

df.dtypes
```

```python
# dataset info

df.info()
```

```python
# 2️⃣ Data Cleaning Operations
# check missing values

df.isnull().sum()
```

```python
# check duplicates

df.duplicated().sum()
```

```python
# remove duplicates (if any)

df = df.drop_duplicates()
```

```python
# 3️⃣ Statistical Analysis
# Descriptive statistics

df.describe()
```

```python
# mean tip

df['tip'].mean()
```

```python
# median total bill

df['total_bill'].median()
```

```python
# Maximum tip

df['tip'].max()
```

```python
# Minimum tip

df['tip'].min()
```

```python
# 4️⃣ GroupBy Operations
# Average tip by gender

df.groupby('sex')['tip'].mean()
```

```python
# Average total bill by day

df.groupby('day')['total_bill'].mean()
```

```python
# Average tip by smoker

df.groupby('smoker')['tip'].mean()
```

```python
# Total revenue by day

df.groupby('day')['total_bill'].sum()
```

```python
# Average tip by time (Lunch/Dinner)

df.groupby('time')['tip'].mean()
```

```python
# 5️⃣ Filtering Operations
# Tips greater than 5

df[df['tip'] > 5]
```

```python
# Non-smokers only

df[df['smoker'] == 'No']
```

```python
# Dinner time data only

df[df['time'] == 'Dinner']
```

```python
# Large group size (>4)

df[df['size'] > 4]
```

```python
# 6️⃣ New Column Creation
# Create tip percentage column

df['tip_percent'] = (df['tip'] / df['total_bill']) * 100
```

```python
# Categorize tip (Low/Medium/High)

df['tip_category'] = pd.cut(df['tip'],
                        bins=[0,3,6,10],
                        labels=['Low','Medium','High'])
```

```python
# 7️⃣ Sorting
# Sort by highest tip

df.sort_values(by='tip', ascending=False)
```

```python
# Sort by total bill

df.sort_values(by='total_bill')
```

```python
# 8️⃣ Value Counts
# Count males & females

df['sex'].value_counts()
```

```python
# Count smokers vs non-smokers

df['smoker'].value_counts()
```

```python
# Count by day

df['day'].value_counts()
```

```python
# 9️⃣ Correlation Analysis
# Correlation matrix

df.corr(numeric_only=True)
```

```python
# Correlation between total_bill and tip

df['total_bill'].corr(df['tip'])
```

```python
# 🔟 Aggregation with Multiple Columns
# Mean tip and bill by day & time

df.groupby(['day','time'])[['total_bill','tip']].mean()
```

```python
# 1️⃣1️⃣ Pivot Table
# Pivot table of average tip

pd.pivot_table(df,
               values='tip',
               index='day',
               columns='sex',
               aggfunc='mean')
```

```python
# 1️⃣2️⃣ Export Processed Data
# Save updated dataset

df.to_csv("tips_processed.csv", index=False)
```
