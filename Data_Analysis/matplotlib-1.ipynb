```python
x = np.arange(1, 11)
y = np.sin(x)
z = np.cos(x)

print(x, y, z, sep = '\n\n')
```

```python
plt.plot(x, y)
```

```python
plt.plot(x, y, label = 'SINQ')
plt.plot(x, z, label = 'COSQ')

plt.show()
```

```python
plt.plot(x, y, label = 'SINQ')
plt.plot(x, z, label = 'COSQ')
plt.legend()
plt.show()
```

```python
plt.plot(x, y, label = 'SINQ', color = 'red')
plt.plot(x, z, label = 'COSQ')
plt.legend()
plt.show()
```

```python
plt.plot(x, y, label = 'SINQ', color = 'red')
plt.plot(x, z, label = 'COSQ')
plt.title("SINQ & COSQ")
plt.xlabel('Data Points')
plt.ylabel('Associate Values')
plt.xticks(x)
plt.yticks(z)
plt.legend()
plt.show()
```

```python
plt.figure(figsize = (8, 4))
plt.plot(x, y, label = 'SINQ', color = 'red')
plt.plot(x, z, label = 'COSQ')
plt.title("SINQ & COSQ")
plt.xlabel('Data Points')
plt.ylabel('Associate Values')
plt.xticks(x)
plt.yticks(z)
plt.legend()
plt.show()
```

```python
# EDA -> explonatory data analysis
# matplotlib
# seaborn
# plotly
```

```python
plt.figure(figsize = (12, 4))

plt.subplot(1, 2, 1)
plt.title("SINQ & COSQ")
plt.plot(x, y, label = 'SINQ', color = 'red')
plt.plot(x, z, label = 'COSQ')
plt.xlabel('Data Points')
plt.ylabel('Associate Values')
plt.grid()

plt.subplot(1, 2, 2)
plt.title("SINQ & COSQ2")
plt.plot(x, y, label = 'SINQ', color = 'green')
plt.plot(x, z, label = 'COSQ', color = 'red')
plt.xlabel('Data Points2')
plt.ylabel('Associate Values2')
plt.subplots_adjust(wspace = 0.5)

plt.grid()
plt.legend()
plt.show()
```

```python
# Sample data
x = np.arange(1, 11)
y = np.sin(x)
z = np.cos(x)

plt.figure(figsize=(10, 8))

for i in range(1, 10):  # 1 → 9
    plt.subplot(3, 3, i)
    plt.plot(x, y, label='SINQ')
    plt.plot(x, z, label='COSQ')
    plt.title(f"Plot {i}")
    plt.xlabel("Data Points")
    plt.ylabel("Values")

plt.legend()
plt.tight_layout()
plt.show()
```
