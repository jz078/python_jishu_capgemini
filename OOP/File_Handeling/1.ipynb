```python
# count number of words into the file
with open('data.txt') as f:
    words = f.read().split(',')
    print(words)
    print(len(words))
f.close()
```

```python
# extract the unique words and count them

with open('data.txt') as f:
    words = f.read().split()
    st1 = set(words)
    print(words)
    print(st1)
f.close()
```

```python
# create a dictionary of words count

with open('data.txt') as f:
    words = f.read().split()

dict1 = {}

for word in words:
    if word in dict1:
        dict1[word] += 1
    else:
        dict1[word] = 1
print(dict1)
f.close()
```

```python
data = {
    "name" : "jishu"
}
```

```python
import json
with open('json1.json', 'w') as f:
    json.dump(data, f) 
```

```python
# read json data

with open('json1.json', 'r') as f:
    print(json.load(f))
```

```python
with open('json1.json', 'r') as f:
    print(json.load(f))
```

```python
with open('json1.json', 'r+') as f:
    json.dump(data, f)
    f.seek(0)
    print(json.load(f))
```
